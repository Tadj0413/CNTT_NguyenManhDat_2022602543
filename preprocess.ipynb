# Tiền Xử Lý Dữ Liệu
Thực hiện xử lý loại bỏ dòng trùng lặp, và xuất ra file dataset để huấn luyện.


```python
import pandas as pd
import warnings
warnings.filterwarnings('ignore')
```

```python
pd.set_option('display.max_columns', 100)
df = pd.read_csv("creditcard.csv")
print("Original shape:", df.shape)
```

```python
fraud_df = df[df['Class'] == 1]
print(fraud_df)
```

```python
df.info()
```

```python
#Check missing values
missing = df.isnull().sum().to_frame().T
print(missing)
```

```python
# Drop duplicates if any
duplicates = df.duplicated().sum()
if duplicates > 0:
    df = df.drop_duplicates()
    print("Dropped", duplicates, "duplicate rows. New shape:", df.shape)
```

```python
# Xuất file dữ liệu đã tiền xử lý
df.to_csv("creditcard_processed.csv", index=False)
print("Saved to creditcard_processed.csv")
```
