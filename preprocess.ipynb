Thử nghiệm loại bỏ ngoại lai dùng IQR

```python
# Hàm loại bỏ ngoại lai bằng phương pháp IQR chỉ dành cho Class = 0
def drop_outliers_iqr(df, column, class_val=0, threshold=1.5):
    class_idx = df[df['Class'] == class_val].index
    v = df.loc[class_idx, column]
    
    Q1 = v.quantile(0.25)
    Q3 = v.quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - threshold * IQR
    upper_bound = Q3 + threshold * IQR
    
    outliers = class_idx[(v < lower_bound) | (v > upper_bound)]
    print(f'\nSố lượng ngoại lai ở cột {column} (Class={class_val}): {len(outliers)}')
    
    df_cleaned = df.drop(outliers)
    return df_cleaned

# Gợi ý: Thường các đặc trưng có tương quan mạnh nhất (âm hoặc dương) với Class sẽ được đem ra xử lý ngoại lai (ví dụ V14, V12, V10)
# Ở đây ta lấy ví dụ với V14:
# df = drop_outliers_iqr(df, 'V14', class_val=0)
# print("Kích thước dữ liệu sau khi xóa ngoại lai:", df.shape)

# df = drop_outliers_iqr(df, 'V17', class_val=0)
# print("Kích thước dữ liệu sau khi xóa ngoại lai:", df.shape)

# df = drop_outliers_iqr(df, 'V12', class_val=0)
# print("Kích thước dữ liệu sau khi xóa ngoại lai:", df.shape)

# df = drop_outliers_iqr(df, 'V10', class_val=0)
# print("Kích thước dữ liệu sau khi xóa ngoại lai:", df.shape)
```

Phân tích hệ số tương quan (Correlation Analysis)
Để biết đặc trưng nào (V1, V2, ..., V28) có tương quan mạnh nhất (âm hoặc dương) với biến mục tiêu `Class`, 
Ta sẽ tính toán ma trận tương quan và trực quan hóa các hệ số này.

```python
# Tính toán ma trận tương quan Pearson
corr_matrix = df.corr()

# Lấy ra mức độ tương quan của từng cột đối chiếu với biến mục tiêu 'Class'
class_corr = corr_matrix['Class'].drop('Class')

# Sắp xếp các hệ số tương quan từ cao xuống thấp
class_corr_sorted = class_corr.sort_values(ascending=False)

# Trực quan hóa mức độ tương quan bằng biểu đồ cột (Bar plot)
plt.figure(figsize=(14, 6))
# Đặt màu đỏ (salmon) cho tương quan âm, màu xanh (skyblue) cho tương quan dương
colors = ['salmon' if c < 0 else 'skyblue' for c in class_corr_sorted.values]

class_corr_sorted.plot(kind='bar', color=colors)
plt.title('Mức độ tương quan của các đặc trưng đối với Class', fontsize=14)
plt.ylabel('Hệ số tương quan (Pearson)')
plt.xlabel('Các đặc trưng')
plt.axhline(0, color='black', linewidth=0.8, linestyle='--')
plt.show()

# Hiển thị Top 4 đặc trưng có tương quan ÂM mạnh nhất (nghịch biến)
print("Top 4 đặc trưng tương quan ÂM (Nghịch biến) mạnh nhất với Class:")
print(class_corr_sorted.tail(4))

print("\nTop 4 đặc trưng tương quan DƯƠNG (Đồng biến) mạnh nhất với Class:")
print(class_corr_sorted.head(4))
```

Vẽ Heatmap thể hiện mức độ quan hệ giữa biến `Class` và các đặc trưng thông qua màu sắc.

```python
# Thiết lập kích thước đồ thị
plt.figure(figsize=(20, 16))

# Tùy chọn 1: Vẽ Heatmap cho toàn bộ ma trận tương quan (30 biến)
# sns.heatmap(corr_matrix, cmap='coolwarm_r', annot=False, linewidths=0.1)
# plt.title('Heatmap cho toàn bộ đặc trưng', fontsize=16)
# plt.show()

# Tùy chọn 2: Vẽ cụ thể cho Heatmap của Top 10 đặc trưng tương quan mạnh nhất (âm hoặc dương) với Class để dễ nhìn (Khuyên dùng)
top_10_features = abs(class_corr).sort_values(ascending=False).head(10).index.tolist()
top_10_features.append('Class') # Thêm Class để đối chiếu

# Lọc dataframe ma trận tương quan chỉ bao gồm các biến này
sub_df = df[top_10_features]
sub_corr = sub_df.corr()

plt.figure(figsize=(10, 8))
# Sử dụng annot=True để in sẵn các con số lên từng ô
sns.heatmap(sub_corr, cmap='coolwarm_r', annot=True, fmt=".2f", linewidths=0.5, vmin=-1, vmax=1)
plt.title('Heatmap bóc tách Top 10 đặc trưng ảnh hưởng mạnh nhất đến Class', fontsize=14)
plt.show()
```

### Phân tích sự mất cân bằng của biến mục tiêu (Class)
Một đặc điểm cốt lõi của bài toán phát hiện gian lận là sự chênh lệch số lượng rất lớn giữa giao dịch bình thường (Class 0) và giao dịch gian lận (Class 1).

```python
# 2. Đếm số lượng và tính tỷ lệ gian lận
class_counts = df['Class'].value_counts()
class_percentages = df['Class'].value_counts(normalize=True).round(4) * 100

print("Số lượng giao dịch:")
print("Bình thường (0):", class_counts[0])
print("Gian lận    (1):", class_counts[1])

print("\nTỷ lệ phần trăm (%):")
print("Bình thường (0):", class_percentages[0], "%")
print("Gian lận    (1):", class_percentages[1], "%")

# Trực quan hóa Bar chart
plt.figure(figsize=(7, 5))
colors = ["#0101DF", "#DF0101"]
sns.countplot(x='Class', data=df, palette=colors)
plt.title('Phân phối giao dịch \n (0: Bình thường || 1: Gian lận)', fontsize=14)
plt.show()
```

Phân bố của Time và Amount
Biểu đồ Histogram để xem xét mức độ trải rộng giá trị của số tiền giao dịch (`Amount`) và thời gian giao dịch (`Time`).

```python
# 3. Trực quan hóa bằng Histogram (Biểu đồ tần suất)
fig, ax = plt.subplots(1, 2, figsize=(18,4))

# Nhận diện tên cột vì có thể bạn đã thay đổi bằng Scaler trước đó
amount_col = 'scaled_amount' if 'scaled_amount' in df.columns else 'Amount'
time_col = 'scaled_time' if 'scaled_time' in df.columns else 'Time'

sns.histplot(df[amount_col].values, bins=50, ax=ax[0], color='r', kde=True)
ax[0].set_title(f'Phân phối tần suất của {amount_col}', fontsize=14)
ax[0].set_xlim([min(df[amount_col].values), max(df[amount_col].values)])

sns.histplot(df[time_col].values, bins=50, ax=ax[1], color='b', kde=True)
ax[1].set_title(f'Phân phối tần suất của {time_col}', fontsize=14)
ax[1].set_xlim([min(df[time_col].values), max(df[time_col].values)])

plt.show()
```

So sánh phân phối của các đặc trưng quan trọng giữa 2 Nhóm (Class 0 vs Class 1)
Biểu đồ **Boxplot** để quan sát sự phân tách này đối với các đại lượng vừa tìm ra có tương quan mạnh nhất với Class (ví dụ: V14, V12, V10, V4, V11).

```python
# 4. Trực quan hóa giá trị bằng Boxplot
features_to_plot = ['V14', 'V12', 'V10', 'V4', 'V11']

# Lọc chỉ vẽ những biểu đồ nào cột tồn tại trong DataFrame để tránh lỗi
existing_features = [f for f in features_to_plot if f in df.columns]

if existing_features:
    fig, axes = plt.subplots(ncols=len(existing_features), figsize=(24, 6))

    # Nếu chỉ có 1 biểu đồ, biến axes sẽ không phải là mảng
    if len(existing_features) == 1:
        axes = [axes]

    for i, feature in enumerate(existing_features):
        sns.boxplot(x="Class", y=feature, data=df, palette=["#0101DF", "#DF0101"], ax=axes[i])
        axes[i].set_title(f'{feature} phân tách theo Class', fontsize=14)
            
    plt.suptitle('Biểu đồ Hộp (Boxplot) so sánh sự phân tách nhóm Bình thường và Gian lận', fontsize=18)
    plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    plt.show()
```

Phân phối Mật độ (KDE) của phần dữ liệu gian lận và bình thường
Tiếp tục vẽ biểu đồ tỷ lệ mật độ (Kernel Density Estimation - KDE) cho cùng các đặc trưng trên để phân tích sâu hơn về không gian điểm dữ liệu có bị đè lên nhau hay không (Distribution Overlap).

```python
import warnings
warnings.filterwarnings("ignore") # Tắt cảnh báo Future Warning của thư viện

# 5. Phân phối mật độ xác suất bằng KDE
if existing_features:
    fig, axes = plt.subplots(ncols=len(existing_features), figsize=(24, 6))
    
    if len(existing_features) == 1:
        axes = [axes]

    for i, feature in enumerate(existing_features):
        sns.kdeplot(df[feature][df.Class == 0], label="Class 0 (Normal)", color="#0101DF", fill=True, ax=axes[i])
        sns.kdeplot(df[feature][df.Class == 1], label="Class 1 (Fraud)", color="#DF0101", fill=True, ax=axes[i])
        
        axes[i].set_title(f'Phân phối KDE của {feature}', fontsize=14)
        axes[i].legend()
            
    plt.suptitle('Biểu đồ Mật độ Xác suất (KDE) bóc tách giao dịch Gian Lận vs Bình Thường', fontsize=18)
    plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    plt.show()
```
